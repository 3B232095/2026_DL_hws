```python
import numpy as np

np.set_printoptions(precision=6, suppress=True)

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    # 每列是一筆資料，對每列的三個類別計算 Softmax
    x = x - np.max(x, axis=1, keepdims=True)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

def cross_entropy_error(y_hat, y):
    # 先計算每筆資料的誤差，再取整批平均
    delta = 1e-7
    individual_loss = -np.sum(y * np.log(y_hat + delta), axis=1)
    return np.mean(individual_loss), individual_loss
```

```python
X = np.array([
    [ 0.5,  1.0, -0.5,  2.0,  1.5, -1.0],
    [ 1.0, -0.5,  0.3,  1.2, -0.7,  0.8],
    [-1.0,  0.2,  1.5, -0.3,  0.7,  1.1],
    [ 0.3,  0.8,  1.0, -1.2,  0.5,  0.4],
    [ 1.2,  0.4, -0.8,  0.6,  1.0, -0.2]
])

# 五筆資料的真實類別依序為第 3、1、2、3、1 類
Y = np.array([
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [1, 0, 0]
])

print("X =\n", X)
print("X shape =", X.shape)
print("\nY =\n", Y)
print("Y shape =", Y.shape)
```

```python
W1 = np.array([
    [ 0.2, -0.1,  0.3,  0.4, -0.2,  0.1],
    [-0.3,  0.5, -0.2,  0.1,  0.4, -0.1],
    [ 0.6,  0.2,  0.1, -0.3,  0.2,  0.5],
    [ 0.1, -0.4,  0.5,  0.2, -0.3,  0.6],
    [-0.2,  0.3, -0.4,  0.5,  0.1, -0.2]
])
b1 = np.array([0.1, -0.2, 0.05, 0.1, -0.1])

W2 = np.array([
    [ 0.2, -0.3,  0.4,  0.1, -0.2],
    [ 0.5,  0.2, -0.1,  0.3,  0.4],
    [-0.4,  0.3,  0.2, -0.2,  0.1],
    [ 0.1,  0.6, -0.3,  0.2, -0.4]
])
b2 = np.array([0.1, -0.1, 0.05, 0.2])

W3 = np.array([
    [ 0.3, -0.2,  0.4,  0.1],
    [-0.5,  0.3,  0.2, -0.4],
    [ 0.2,  0.5, -0.3,  0.6],
    [ 0.4, -0.1,  0.5, -0.2]
])
b3 = np.array([0.1, 0.0, -0.1, 0.2])

W4 = np.array([
    [ 0.2, -0.3,  0.5,  0.1],
    [-0.4,  0.6, -0.2,  0.3],
    [ 0.3,  0.1, -0.4,  0.5]
])
b4 = np.array([0.1, -0.1, 0.2])

for name, value in [
    ("W1", W1), ("b1", b1), ("W2", W2), ("b2", b2),
    ("W3", W3), ("b3", b3), ("W4", W4), ("b4", b4)
]:
    print(f"{name} =\n{value}")
    print(f"{name} shape = {value.shape}, 數值個數 = {value.size}\n")
```

```python
# 第一隱藏層：6 -> 5
z1 = X @ W1.T + b1
h1 = relu(z1)

# 第二隱藏層：5 -> 4
z2 = h1 @ W2.T + b2
h2 = relu(z2)

# 第三隱藏層：4 -> 4
z3 = h2 @ W3.T + b3
h3 = sigmoid(z3)

# 輸出層：4 -> 3
z4 = h3 @ W4.T + b4
y_hat = softmax(z4)

for name, value in [
    ("z1", z1), ("h1", h1), ("z2", z2), ("h2", h2),
    ("z3", z3), ("h3", h3), ("z4", z4), ("y_hat", y_hat)
]:
    print(f"{name} =\n{value}")
    print(f"{name} shape = {value.shape}\n")
```

```python
average_loss, individual_losses = cross_entropy_error(y_hat, Y)

print("每筆資料的 y_hat：\n", y_hat)
print("\n真實答案 Y：\n", Y)
print("\n每筆資料的 Cross Entropy Error：")
print(individual_losses)
print("\n五筆資料的平均 Cross Entropy Error =", average_loss)

predicted_classes = np.argmax(y_hat, axis=1) + 1
true_classes = np.argmax(Y, axis=1) + 1
print("\n預測類別 =", predicted_classes)
print("真實類別 =", true_classes)
```
