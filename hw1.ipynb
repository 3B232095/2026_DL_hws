```python
import numpy as np

# ==
# 1. 定義激發函數
# ==

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    # 避免 exp 數值過大
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)


# ==
# 2. 定義 Cross Entropy Error
# ==

def cross_entropy_error(y_hat, y):
    delta = 1e-7
    return -np.sum(y * np.log(y_hat + delta))


# ==
# 3. 輸入資料
# ==

# x 有 6 個特徵值
x = np.array([0.5, 1.0, -0.5, 2.0, 1.5, -1.0])

print("x =", x)
print("x shape =", x.shape)


# ==
# 4. 第一隱藏層
#    6 -> 5
#    ReLU
# ==

W1 = np.array([
    [0.2, -0.1, 0.3, 0.4, -0.2, 0.1],
    [-0.3, 0.5, -0.2, 0.1, 0.4, -0.1],
    [0.6, 0.2, 0.1, -0.3, 0.2, 0.5],
    [0.1, -0.4, 0.5, 0.2, -0.3, 0.6],
    [-0.2, 0.3, -0.4, 0.5, 0.1, -0.2]
])

b1 = np.array([0.1, -0.2, 0.05, 0.1, -0.1])

z1 = np.dot(W1, x) + b1
h1 = relu(z1)

print("\n========== 第一隱藏層 ==========")
print("W1 =")
print(W1)
print("W1 shape =", W1.shape)
print("W1 個數 =", W1.size)

print("b1 =", b1)
print("b1 shape =", b1.shape)
print("b1 個數 =", b1.size)

print("z1 =", z1)
print("z1 shape =", z1.shape)

print("h1 =", h1)
print("h1 shape =", h1.shape)


# ==
# 5. 第二隱藏層
#    5 -> 4
#    ReLU
# ==

W2 = np.array([
    [0.2, -0.3, 0.4, 0.1, -0.2],
    [0.5, 0.2, -0.1, 0.3, 0.4],
    [-0.4, 0.3, 0.2, -0.2, 0.1],
    [0.1, 0.6, -0.3, 0.2, -0.4]
])

b2 = np.array([0.1, -0.1, 0.05, 0.2])

z2 = np.dot(W2, h1) + b2
h2 = relu(z2)

print("\n========== 第二隱藏層 ==========")
print("W2 =")
print(W2)
print("W2 shape =", W2.shape)
print("W2 個數 =", W2.size)

print("b2 =", b2)
print("b2 shape =", b2.shape)
print("b2 個數 =", b2.size)

print("z2 =", z2)
print("z2 shape =", z2.shape)

print("h2 =", h2)
print("h2 shape =", h2.shape)


# ==
# 6. 第三隱藏層
#    4 -> 4
#    Sigmoid
# ==

W3 = np.array([
    [0.3, -0.2, 0.4, 0.1],
    [-0.5, 0.3, 0.2, -0.4],
    [0.2, 0.5, -0.3, 0.6],
    [0.4, -0.1, 0.5, -0.2]
])

b3 = np.array([0.1, 0.0, -0.1, 0.2])

z3 = np.dot(W3, h2) + b3
h3 = sigmoid(z3)

print("\n========== 第三隱藏層 ==========")
print("W3 =")
print(W3)
print("W3 shape =", W3.shape)
print("W3 個數 =", W3.size)

print("b3 =", b3)
print("b3 shape =", b3.shape)
print("b3 個數 =", b3.size)

print("z3 =", z3)
print("z3 shape =", z3.shape)

print("h3 =", h3)
print("h3 shape =", h3.shape)


# ==
# 7. 輸出層
#    4 -> 3
#    Softmax
# ==

W4 = np.array([
    [0.2, -0.3, 0.5, 0.1],
    [-0.4, 0.6, -0.2, 0.3],
    [0.3, 0.1, -0.4, 0.5]
])

b4 = np.array([0.1, -0.1, 0.2])

z4 = np.dot(W4, h3) + b4
y_hat = softmax(z4)

print("\n========== 輸出層 ==========")
print("W4 =")
print(W4)
print("W4 shape =", W4.shape)
print("W4 個數 =", W4.size)

print("b4 =", b4)
print("b4 shape =", b4.shape)
print("b4 個數 =", b4.size)

print("z4 =", z4)
print("z4 shape =", z4.shape)

print("y_hat =", y_hat)
print("y_hat shape =", y_hat.shape)


# ==
# 8. 答案
# ==

y = np.array([0, 0, 1])

print("\n========== Cross Entropy Error ==========")
print("y =", y)

loss = cross_entropy_error(y_hat, y)

print("Cross Entropy Error =", loss)
```

```python
import numpy as np
import gradio as gr


# ==
# 1. 定義激發函數
# ==

def relu(x):
    return np.maximum(0, x)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def softmax(x):
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)


# ==
# 2. 定義 DNN 函數
# ==

def dnn(x):

    # 將輸入資料轉成 numpy array
    x = np.array(x)

    # --
    # 第一隱藏層：6 -> 5
    # ReLU
    # --

    W1 = np.array([
        [0.2, -0.1, 0.3, 0.4, -0.2, 0.1],
        [-0.3, 0.5, -0.2, 0.1, 0.4, -0.1],
        [0.6, 0.2, 0.1, -0.3, 0.2, 0.5],
        [0.1, -0.4, 0.5, 0.2, -0.3, 0.6],
        [-0.2, 0.3, -0.4, 0.5, 0.1, -0.2]
    ])

    b1 = np.array([0.1, -0.2, 0.05, 0.1, -0.1])

    z1 = np.dot(W1, x) + b1
    h1 = relu(z1)


    # --
    # 第二隱藏層：5 -> 4
    # ReLU
    # --

    W2 = np.array([
        [0.2, -0.3, 0.4, 0.1, -0.2],
        [0.5, 0.2, -0.1, 0.3, 0.4],
        [-0.4, 0.3, 0.2, -0.2, 0.1],
        [0.1, 0.6, -0.3, 0.2, -0.4]
    ])

    b2 = np.array([0.1, -0.1, 0.05, 0.2])

    z2 = np.dot(W2, h1) + b2
    h2 = relu(z2)


    # --
    # 第三隱藏層：4 -> 4
    # Sigmoid
    # --

    W3 = np.array([
        [0.3, -0.2, 0.4, 0.1],
        [-0.5, 0.3, 0.2, -0.4],
        [0.2, 0.5, -0.3, 0.6],
        [0.4, -0.1, 0.5, -0.2]
    ])

    b3 = np.array([0.1, 0.0, -0.1, 0.2])

    z3 = np.dot(W3, h2) + b3
    h3 = sigmoid(z3)


    # --
    # 輸出層：4 -> 3
    # Softmax
    # --

    W4 = np.array([
        [0.2, -0.3, 0.5, 0.1],
        [-0.4, 0.6, -0.2, 0.3],
        [0.3, 0.1, -0.4, 0.5]
    ])

    b4 = np.array([0.1, -0.1, 0.2])

    z4 = np.dot(W4, h3) + b4

    y_hat = softmax(z4)


    # 回傳預測結果
    return y_hat


# ==
# 3. 上一題的 x
# ==

x = np.array([0.5, 1.0, -0.5, 2.0, 1.5, -1.0])

y_hat = dnn(x)

print("輸入 x：")
print(x)

print("\ndnn(x) 的結果 y_hat：")
print(y_hat)

print("\ny_hat shape：")
print(y_hat.shape)


# ==
# 4. 找出機率最高的類別
# ==

class_number = np.argmax(y_hat) + 1

print("\n機率最高的類別：")
print(class_number)


# ==========================================
# 5. Gradio 互動介面
# ==========================================

def predict(x1, x2, x3, x4, x5, x6):

    # 將 6 個輸入值組成 x
    x = np.array([x1, x2, x3, x4, x5, x6])

    # 使用 DNN 計算
    y_hat = dnn(x)

    # 找出機率最高的類別
    result = np.argmax(y_hat) + 1

    return f"第 {result} 類別"


# 建立 Gradio 介面
demo = gr.Interface(
    fn=predict,
    inputs=[
        gr.Number(label="特徵值 x1"),
        gr.Number(label="特徵值 x2"),
        gr.Number(label="特徵值 x3"),
        gr.Number(label="特徵值 x4"),
        gr.Number(label="特徵值 x5"),
        gr.Number(label="特徵值 x6")
    ],
    outputs=gr.Textbox(label="預測結果"),
    title="DNN 三分類預測",
    description="請輸入 6 個特徵值"
)

# 啟動介面
demo.launch()
```
